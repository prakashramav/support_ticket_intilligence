# Data Exploration
Let's explore the support tickets dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

df = pd.read_csv("../data/raw/tickets.csv")
df.head()
```

```python
print(f"Shape: {df.shape}")
print(df.info())
print(df.isnull().sum())
```

```python
print(df.columns.tolist())
```

```python
print(df.shape)
print(df.head())
```

```python
for i, col in enumerate(df.columns):
    print(i, repr(col))
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
print(df.columns.tolist())
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(data=df, y="ticket_type")
plt.title("Category Distribution")
plt.tight_layout()
plt.savefig("../reports/figures/category_dist.png")
plt.show()
```

```python
if "ticket_priority" in df.columns:
    plt.figure(figsize=(8, 4))
    sns.countplot(data=df, x="ticket_priority")
    plt.title("Priority Distribution")
    plt.savefig("../reports/figures/priority_dist.png")
    plt.show()
```
