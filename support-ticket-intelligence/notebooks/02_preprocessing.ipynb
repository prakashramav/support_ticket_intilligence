# Preprocessing
Clean the text data and prepare it for feature engineering.

```python
import pandas as pd
import sys
sys.path.append("..")
from src.data.preprocessing import clean_dataset, TextPreprocessor

df = pd.read_csv("../data/raw/tickets.csv")
print("Original shape:", df.shape)
```
